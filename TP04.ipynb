Summary of 4 principles class in OOP

OOP known as Object Oriented Programming Language. Moreover, it has four main principles such as Encapsulation, Abstraction, Inheritance, and Polymorphism.

a). Encapsulation
We use encapsulation for wrapping data and methods into a single unit(class). Addictionally, for protecting data from direct external modification.

b). Abstraction
We use abstraction for hiding complex implementation details. Also, for showing only essential features to the user.

c). Inheritance
We use inheritance for the child class to inherits the proporties and methods from a parent class. Moreover, it also promotes code reuse either.

d). Polymorphism
We use polymorphism for same method name that behave differently depending on the object. 


```python
#Encapsulation
class Car:
    def __init__(self):
        self.__max_speed = 200

    def get_max_speed(self):
        return self.__max_speed

car = Car()
print(car.get_max_speed())
```

```python
#Abstraction
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass
```

```python
#Inheritance
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        print("dog barks")

animal = Animal()
animal.speak()
dog1 = Dog()
dog1.speak()

    
```

```python
#Polymorphism
class Dog:
    def speak(self):
        print("Wolf")

class Cat:
    def speak(self):
        print("Meow")

animals = [Dog(), Cat()]

for animal in animals:
    animal.speak()
```

```python
#Homework
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self._price = price

    def get_price(self):                 #I want to access into private data
        return self._price

    def set_price(self, new_price):      #I want to modify the values 
        if new_price < 0:
            print("Price cannot be negative!")
        else:
            self._price = new_price

    def show_info(self):
        print(self.title, "by", self.author, "- $", self._price)


class EBook(Book):
    def __init__(self, title, author, price, file_size):
        super().__init__(title, author, price)
        self.file_size = file_size

    def show_info(self):
        super().show_info()
        print("File size:", self.file_size, "MB")


book = Book("1984", "George Orwell", 12.99)
book.show_info()

book.set_price(15.50)
print("New price:", book.get_price())

book.set_price(-5)

ebook = EBook("1984", "George Orwell", 7.99, 2.5)
ebook.show_info()
```
